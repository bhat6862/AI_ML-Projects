```python
import getpass

# Set the "stored" password and limit attempts
stored_password = "Secure_Python_2026!"
attempts = 3

print("--- Secure Login System ---")

while attempts > 0:
    # 1. Use getpass to mask the input (password won't show as you type)
    user_input = getpass.getpass(f"Enter your password ({attempts} attempts remaining): ")

    # 2. Check for the correct password
    if user_input == stored_password:
        print("Verification Successful. Access Granted.")
        break
    
    # 3. Add complexity feedback if they are setting a new password 
    # (Or simply provide a hint for the existing one)
    else:
        attempts -= 1
        if attempts > 0:
            print("Access Denied. Please try again.")
            
            # Simple complexity hint
            if len(user_input) < 8:
                print("Hint: The password is at least 8 characters long.")
        else:
            print("\nCritical Error: Too many failed attempts. System Locked.")
```
